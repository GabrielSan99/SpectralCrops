```python
import matplotlib.pyplot as plt 
import matplotlib.image as img
import numpy as np
```

```python
image_blue = img.imread('/home/sanches/Documents/SpectralCrops/spectral_project/captures_test/captures_2025-08-01_16-27-47/blue_2025-08-01_16-27-47.png') 
image_red = img.imread('/home/sanches/Documents/SpectralCrops/spectral_project/captures_test/captures_2025-08-01_16-27-47/red_2025-08-01_16-27-47.png') 
image_yellow = img.imread('/home/sanches/Documents/SpectralCrops/spectral_project/captures_test/captures_2025-08-01_16-27-47/yellow_2025-08-01_16-27-47.png')
image_white = img.imread('/home/sanches/Documents/SpectralCrops/spectral_project/captures_test/captures_2025-08-01_16-27-47/white_2025-08-01_16-27-47.png')

# nipy_spectral is indicate to spectral_images and viridis is the default method
plt.set_cmap('nipy_spectral')

print("Dimensão Imagem Azul: ", image_blue.shape)
print("Dimensão Imagem Vermelho: ", image_red.shape)
print("Dimensão Imagem Amarelo: ", image_yellow.shape)
print("Dimensão Imagem Branca: ", image_white.shape)

# Add the first image to the figure (top-left position)
plt.subplot(2, 2, 1)  # 2 rows, 2 columns, first position
plt.imshow(image_blue)  
plt.axis('off')  # Hide the axis labels
plt.title("Imagem Azul") 

# Add the second image to the figure (top-right position)
plt.subplot(2, 2, 2)  # 2 rows, 2 columns, second position
plt.imshow(image_red)  
plt.axis('off')  # Hide the axis labels
plt.title("Imagem Vermelho") 

# Add the third image to the figure (bottom-left position)
plt.subplot(2, 2, 3)  # 2 rows, 2 columns, third position
plt.imshow(image_yellow) 
plt.axis('off')  # Hide the axis labels
plt.title("Imagem Amarelo")  

# Add the fourth image to the figure (bottom-right position)
plt.subplot(2, 2, 4)  # 2 rows, 2 columns, fourth position
plt.imshow(image_white)  
plt.axis('off')  # Hide the axis labels
plt.title("Imagem Branco")  
```
