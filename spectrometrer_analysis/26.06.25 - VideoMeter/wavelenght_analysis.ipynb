```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.io as pio
import numpy as np
import math
import os
```

```python
spectral_dict = {
    'UVA': {'wavelength': 365, 'file': ''},
    'Violet': {'wavelength': 405, 'file': ''},
    'Indigo': {'wavelength': 430, 'file': ''},
    'Blue_1': {'wavelength': 450, 'file': ''},
    'Blue_2': {'wavelength': 470, 'file': ''},
    'Azure': {'wavelength': 490, 'file': ''},
    'Cyan': {'wavelength': 515, 'file': ''},
    'Green': {'wavelength': 540, 'file': ''},
    'Yellow': {'wavelength': 570, 'file': ''},
    'Amber': {'wavelength': 590, 'file': ''},
    'Red_1': {'wavelength': 630, 'file': ''},
    'Red_2': {'wavelength': 645, 'file': ''},
    'Red_3': {'wavelength': 660, 'file': ''},
    'Red_4': {'wavelength': 690, 'file': ''},
    'NIR_1': {'wavelength': 780, 'file': ''},
    'NIR_2': {'wavelength': 850, 'file': ''},
    'NIR_3': {'wavelength': 880, 'file': ''},
    'NIR_4': {'wavelength': 940, 'file': ''},
    'NIR_5': {'wavelength': 970, 'file': ''}
}
```

```python
df = pd.read_csv(r"VideoMeter_lights_off\capture00000.txt",
                 sep='\t', header=None, names=['spectro', 'intensity'])

df
```

```python
# Transformar a coluna spectro em inteiros para coincidir com as chaves do dicionário
df['spectro'] = df['spectro'].str.replace(',', '.')
df['spectro'] = pd.to_numeric(df['spectro'], errors='coerce')  # float
df['spectro'] = df['spectro'].round().astype('Int64')
df
```

# Lights ON

```python
folder_path = 'VideoMeter_lights_on'
```

```python
for name, spectro in spectral_dict.items():

    files = [os.path.join(folder_path, nome) for nome in os.listdir(folder_path)]

    aux_dict = {}
    for file in files:
      df = pd.read_csv(file, sep='\t', header=None, names=['spectro', 'intensity'])

      # Transformar a coluna spectro em inteiros para coincidir com as chaves do dicionário
      df = df.replace(',', '.', regex=True)
      df = df.apply(pd.to_numeric, errors='coerce')
      df['spectro'] = df['spectro'].round().astype('Int64')

      df_filtered = df[df['spectro'] == spectro['wavelength']]
      aux_dict[file] = round(df_filtered['intensity'].mean(), 3)

    max_file = max(aux_dict, key=aux_dict.get)
    max_value = aux_dict[max_file]

    print(f'{name}: biggest intensity = {max_value} in the file {os.path.basename(max_file)}')
    spectral_dict[name]['file'] = os.path.basename(max_file)


```

```python
spectral_dict
```

```python
# Corresponds to blue_2 wavelenght biggest intensity
file_to_plot = "VideoMeter_lights_on/capture00069.txt"


df_plot = pd.read_csv(file_to_plot, sep='\t', header=None, names=['spectro', 'intensity'])
df_plot = df_plot.replace(',', '.', regex=True)
df_plot = df_plot.apply(pd.to_numeric, errors='coerce')

fig = px.line(df_plot, x='spectro', y='intensity', title=f'Intensity vs Wavelength for {os.path.basename(file_to_plot)}')
fig.update_layout(
    xaxis_title='Wavelength (spectro)',
    yaxis_title='Intensity',
    hovermode='x unified' # Optional: show tooltips for all lines at a given x-value
)

pio.renderers.default = "notebook_connected"
fig.show()
```
